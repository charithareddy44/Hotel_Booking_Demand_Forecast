```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/hotel_bookings_raw.csv")
df.shape
```

```python
duplicates_before = df.duplicated().sum()
df = df.drop_duplicates()
duplicates_after = df.duplicated().sum()

duplicates_before, duplicates_after
```

```python
df['adr'] = df['adr'].fillna(df['adr'].median())
```

```python
df['customer_type'] = df['customer_type'].fillna(df['customer_type'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df['total_nights'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']
```

```python
df.to_csv("../data/processed/hotel_bookings_clean.csv", index=False)
```
